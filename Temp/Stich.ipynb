```python
import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt
```

```python

def stitch_csv_files(input_folder, output_filepath):
    all_dfs = []
    for filename in os.listdir(input_folder):
        if filename.endswith(".csv"):
            filepath = os.path.join(input_folder, filename)
            try:
                df = pd.read_csv(filepath)
                all_dfs.append(df)
            except Exception as e:
                print(f"Error reading {filename}: {e}")
    
    if all_dfs:
        combined_df = pd.concat(all_dfs, ignore_index=True)
        combined_df.to_csv(output_filepath, index=False)
        print(f"Successfully stitched {len(all_dfs)} CSV files into {output_filepath}")
    else:
        print("No CSV files found to stitch.")
```

```python

adjusted_data_folder = 'C:/Users/Dhile/Projects/Stock/Temp'
output_stitched_file = 'C:/Users/Dhile/Projects/Stock/adani_unseen_adjusted.csv'

stitch_csv_files(adjusted_data_folder, output_stitched_file)
```
