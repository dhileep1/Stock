```python
df['future_return_7d'] = (df['Close'].shift(-7) / df['Close'] - 1) * 100
df
```

```python
def classify_return(x):
    if x <= -10: return 0
    elif -10 < x <= -2: return 1
    elif -2 < x < 2: return 2
    elif 2 <= x < 10: return 3
    else: return 4
```

```python
import sys
print(sys.executable)
```

```python
import ta
```

```python

# --- Target Variable ---
# Assuming 'future_return_7d' already exists
df['target_class'] = df['future_return_7d'].apply(classify_return)

# --- Core Returns ---
df['return_1d'] = df['Close'].pct_change() * 100
df['return_5d'] = df['Close'].pct_change(5) * 100

# --- Momentum Indicators ---
df['rsi_14'] = ta.momentum.RSIIndicator(close=df['Close'], window=14).rsi()
df['roc_10'] = ta.momentum.ROCIndicator(close=df['Close'], window=10).roc()
df['cci_20'] = ta.trend.CCIIndicator(
    high=df['High'], low=df['Low'], close=df['Close'], window=20
).cci()

# --- Trend Indicators ---
ema_12 = ta.trend.EMAIndicator(close=df['Close'], window=12)
ema_26 = ta.trend.EMAIndicator(close=df['Close'], window=26)
macd = ta.trend.MACD(close=df['Close'])

df['ema_12'] = ema_12.ema_indicator()
df['ema_26'] = ema_26.ema_indicator()
df['macd'] = macd.macd()
df['signal'] = macd.macd_signal()
df['macd_hist'] = macd.macd_diff()

# --- Volatility Indicators ---
atr = ta.volatility.AverageTrueRange(high=df['High'], low=df['Low'], close=df['Close'], window=14)
bb = ta.volatility.BollingerBands(close=df['Close'], window=20)

df['atr_14'] = atr.average_true_range()
df['bb_high'] = bb.bollinger_hband()
df['bb_low'] = bb.bollinger_lband()
df['bb_width'] = (df['bb_high'] - df['bb_low']) / df['Close']

# --- Volume Indicators ---
obv = ta.volume.OnBalanceVolumeIndicator(close=df['Close'], volume=df['Volume'])
mfi = ta.volume.MFIIndicator(high=df['High'], low=df['Low'], close=df['Close'], volume=df['Volume'], window=14)

df['obv'] = obv.on_balance_volume()
df['mfi'] = mfi.money_flow_index()

# --- Rolling Statistics ---
df['roll_mean_5'] = df['Close'].rolling(window=5).mean()
df['roll_std_5'] = df['Close'].rolling(window=5).std()
df['roll_mean_20'] = df['Close'].rolling(window=20).mean()
df['roll_std_20'] = df['Close'].rolling(window=20).std()

# --- Lagged Returns ---
for lag in [1, 3, 5]:
    df[f'return_lag_{lag}'] = df['return_1d'].shift(lag)

# --- Clean up NaNs created by indicators ---
df = df.dropna().reset_index()
```

```python
df['adx'] = ta.trend.ADXIndicator(df['High'], df['Low'], df['Close']).adx()
df['stoch_k'] = ta.momentum.StochasticOscillator(df['High'], df['Low'], df['Close']).stoch()
df['stoch_d'] = ta.momentum.StochasticOscillator(df['High'], df['Low'], df['Close']).stoch_signal()
df['williams_r'] = ta.momentum.WilliamsRIndicator(df['High'], df['Low'], df['Close']).williams_r()
donchian = ta.volatility.DonchianChannel(df['High'], df['Low'], df['Close'])
df['donchian_h'] = donchian.donchian_channel_hband()
df['donchian_l'] = donchian.donchian_channel_lband()
df['vol_sma_20'] = df['Volume'].rolling(20).mean()
df['vol_ratio'] = df['Volume'] / df['vol_sma_20']
df['vwap_dev'] = (df['Close'] - df['VWAP']) / df['VWAP'] * 100

```

```python
df['Date'] = pd.to_datetime(df['Date'])

# Extract month and day
df['month'] = df['Date'].dt.month
df['day'] = df['Date'].dt.day
df['weekday_num'] = df['Date'].dt.weekday
```

```python
# ✅ Export to CSV
output_path = "feature_engineered_stock_data.csv"
df.to_csv(output_path, index=False)

print(f"✅ File successfully saved as: {output_path}")
print(f"📈 Total rows: {len(df)}, columns: {len(df.columns)}")
```
